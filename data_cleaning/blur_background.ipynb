```python
import cv2
import pixellib
import matplotlib.pyplot as plt

from pixellib.tune_bg import alter_bg
```

```python
# load original image

original_image = cv2.imread("path_to_image")
original_image_in_rgb = cv2.cvtColor(original_image, cv2.COLOR_BGR2RGB)
```

```python
# blur background

blur_background = alter_bg(model_type = "pb")
blur_background.load_pascalvoc_model("/content/xception_pascalvoc.pb")
new_image = blur_background.blur_bg(image_path = "path_to_image", extreme = True)
new_image = cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB)
```

```python
# view output of blurring process compared with original image

plt.figure(figsize = (16, 12))

plt.subplot(121)
plt.imshow(original_image_in_rgb)
plt.title('original_image')

plt.subplot(122)
plt.imshow(new_image)
plt.title('Blurred Image')
```
